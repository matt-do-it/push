# Sample integration of push framework

```r
suppressWarnings(suppressMessages({
    library(IRdisplay)
    library(pushwidgets)
    library(tidyverse)}))

options(htmlwidgets.TOJSON_ARGS = list(auto_unbox = FALSE))

df <- data.frame(date = c("2024-07-01", "2024-07-01", "2024-04-01", "2024-04-01", "2024-01-01", "2024-01-01"), 
                 group = c("Group A", "Group B", "Group A", "Group B", "Group A", "Group B"), 
                 value = c(500, 700, 200, 300, 100, 700))

widget <- pushwidgets::push_summary(df, 
                                    list(title = jsonlite::unbox("Example"), 
                                         display = jsonlite::unbox("isoquarter")), 
                                    groupColumns = c('group', 'date'),
                                    rollup = list(value = jsonlite::unbox('op.sum(d.value)')), 
                                    width = 400,
                                    height = 300)

IRdisplay::display(widget)
```
